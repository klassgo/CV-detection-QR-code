```python
import ultralytics

print(ultralytics.__version__)
ultralytics.checks()
```

```python
from pathlib import Path

DATA_YAML = Path("/kaggle/input/datasets/klassgo/cr-code-detection/data/data.yaml")

print(DATA_YAML.read_text())
```

```python
from ultralytics import YOLO

model = YOLO("yolo26m.pt")
model.info()
```

```python

for i in range(torch.cuda.device_count()):
    print(i, torch.cuda.get_device_name(i))
```

```python
results = model.train(
    data=str(DATA_YAML),
    epochs=30,
    imgsz=1024,
    batch=8,
    device=[0, 1],
    workers=4,
    patience=8,
    amp=True,
    pretrained=True,
    plots=True,
    save=True,
    project="/kaggle/working/runs",
    name="yolo26m_qr_1024"
)
```
